# 불용어 & 정수인덱스

**불용어(Stopword)** <br>
갖고 있는 데이터에서 유의미한 단어 토큰만 선별하기 위해서는 큰 의미가 없는 단어 토큰을 제거하는 작업이 필요
<br>
<br>
즉, 텍스트에 자주 등장하지만 분석에는 큰 도움이 되지 않는 단어

불용어 예시<br>
- 영어: I, my, me, over 등<br>
- 한국어: 조사, 접미사 등


```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
# NLTK에서 불용어 확인하기(영어)
stop_words_list = stopwords.words('english')
print('불용어 개수 :', len(stop_words_list))
print('불용어 10개 출력 :',stop_words_list[:10])
```

```python
# NLTK를 이용한 불용어 제거(영어)
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
example = "Family is not an important thing. It's everything."
```

```python
stop_words = set(stopwords.words('english'))
word_tokens = word_tokenize(example)
```

```python
result = []
for word in word_tokens:
    if word not in stop_words:
        result.append(word)
print('불용어 제거 전 :', word_tokens)
print('불용어 제거 후 :', result)
```

## 정수 인코딩

**정수 인코딩(Integer Encoding)**
- 자연어 처리에서는 텍스트를 숫자로 바꾸는 여러 기법을 사용
    - 그 첫 단계 중 하나가→ 각 단어를 고유한 정수(인덱스)에 매칭하는
전처리 작업
- 예시
    - 단어가 5,000개 있다면→ 각 단어에 1 ~ 5,000까지의 고유 번호 부여
    - 예: book → 150, dog → 171, love → 192, books → 212
- 인덱스 부여 기준
    - 랜덤 부여도 가능하지만 보통은 단어 등장 빈도수를 기준으로 정렬한
뒤 빈도수가 높은 단어부터 낮은 번호를 부여

- 정수 인코딩이 필요한 이유
    -  정수 인코딩은 이후에 배우게 될 원-핫 인코딩, 워드 임베딩(Word Embedding) 등의 수치화 기법의 기본 전처리 단계
- 정수 인코딩 방법(한가지 예시)
    - 텍스트에서 단어들을 수집해 단어 집합(vocabulary) 생성
    - 단어를 빈도수 순으로 정렬
    - 빈도수가 높은 단어부터 1, 2, 3, ... 순서로 인덱스 부여

### dictionary 자료형으로 정수 인코딩

```python
# 예제 텍스트 및 문장 
import nltk
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
raw_text = "A barber is a person. a barber is good person. a barber is huge person. he Knew A Secret! The Secret He Kept is huge secret. Huge secret. His barber kept his word. a barber kept his word. His barber kept his secret. But keeping and keeping such a huge secret to himself was driving the barber crazy. the barber went up a huge mountain."
```

```python
sentences = nltk.sent_tokenize(raw_text)
print(sentences)
```

```python
# 단어 토큰화 + 전처리 과정
vocab = {}
preprocessed_sentences = []
stop_words = set(stopwords.words('english'))
```

```python
for sentence in sentences:
    tokenized_sentence = word_tokenize(sentence) # 단어 토큰화
    result = []

    for word in tokenized_sentence:
        word = word.lower() # 1) 소문자화
        if word not in stop_words: # 2) 불용어 제거
            if len(word) > 2: # 3) 길이 2 이하 단어 제거
                result.append(word)
                if word not in vocab:
                    vocab[word] = 0
                vocab[word] += 1
    preprocessed_sentences.append(result)
```

```python
print(preprocessed_sentences)
```

```python
# 단어 사전(vocab) 생성 및 빈도수 계산
print('단어 집합 :',vocab)
```

```python
# 'barber'라는 단어의 빈도수 출력
print(vocab["barber"])
```

```python
# 빈도수 기준 정렬 → 정수 인덱스 부여
vocab_sorted = sorted(vocab.items(), key = lambda x:x[1],
reverse = True) # 빈도수 기준으로 정렬
print(vocab_sorted)
```

```python
# 빈도수 기준 정렬 → 정수 인덱스 부여
# 빈도수가 높은 단어부터 → 1, 2, 3, ... 순으로 정수 인덱스를 부여
word_to_index = {}
i = 0
for (word, frequency) in vocab_sorted :
    if frequency > 1 : # 빈도수가 1보다 크지 않은 단어는 제외
        i = i + 1
        word_to_index[word] = i
print(word_to_index)
```

```python
# 상위 n개 단어만 사용
# 실제 자연어 처리에서는 텍스트에 등장하는 모든 단어를 사용하는 것보다 빈도수가 가장 높은 n개의 단어만 사용하는 경우가 많음
vocab_size = 5
# 인덱스가 5 초과인 단어 제거
words_frequency = [word for word, index in
    word_to_index.items() if index >= vocab_size + 1]
# 해당 단어에 대한 인덱스 정보를 삭제
for w in words_frequency:
    del word_to_index[w]
print(word_to_index)
```

**Out-Of-Vocabulary(OOV) 문제**
- word_to_index를 사용하여 preprocessed_sentences의 각 단어를
정수로 인코딩
- 예시
    - 첫 번째 문장: ['barber', 'person'] → [1, 5]
    - 두 번째 문장: ['barber', 'good', 'person’]
    - 여기서 'good'은 더 이상 word_to_index에 <u>존재하지 않음</u>
- **단어 집합에 없는 단어가 발생하는 상황을 Out-Of-Vocabulary (OOV)
문제라고 부름**
    - 줄여서 <u>OOV 문제</u>라고도 함



**OOV 토큰 추가하기**
- 사전에 없는 모든 단어는
    - → 'OOV'라는 특별한 토큰의 인덱스로 인코딩
- 이렇게 하면단어 집합 밖의 단어들도 하나의 공통 번호로 표현 가능

```python
word_to_index['OOV'] = len(word_to_index) + 1
print(word_to_index)
```

```python
# 문장을 정수 시퀀스로 인코딩
# 상위 5개 단어 + OOV 토큰만 사용하여 표현
encoded_sentences = []
for sentence in preprocessed_sentences:
    encoded_sentence = []
    for word in sentence:
        try:
            # 단어 집합에 있는 단어라면 해당 단어의 정수를 리턴.
            encoded_sentence.append(word_to_index[word])
        except KeyError:
            # 만약 단어 집합에 없는 단어라면 'OOV'의 정수를 리턴.
            encoded_sentence.append(word_to_index['OOV'])
    encoded_sentences.append(encoded_sentence)
print(encoded_sentences)  # OOV:6을 확인할 수 있음
```

### Counter를 이용한 정수 인코딩

```python
# Counter를 이용한 단어 빈도수 계산
# 준비된 데이터 preprocessed_sentences : 단어 토큰화와 전처리가 끝난 문장 리스트
print(preprocessed_sentences)
```

```python
# 모든 단어를 하나의 리스트로 만들기
# 준비된 데이터 preprocessed_sentences : 단어 토큰화와 전처리가 끝난 문장 리스트

# words = np.hstack(preprocessed_sentences)으로도 수행 가능.
all_words_list = sum(preprocessed_sentences, []) # 리스트끼리 덧셈하면 리스트가 합쳐짐
print(all_words_list)
```

```python
# Counter로 단어 빈도수 계산
from collections import Counter
vocab = Counter(all_words_list)
print(vocab)

# 단어를 키(key)로, 단어에 대한 빈도수가 값(value)으로 저장
# vocab에 단어를 입력하면 빈도수를 리턴

# 'barber'라는 단어의 빈도수 출력
print(vocab["barber"])
```

```python
# 상위 n개 단어만 선택하기 (most_common)
# most_common(n): 등장 빈도수가 높은 상위 n개 단어를 리스트로 반환
vocab_size = 5
vocab = vocab.most_common(vocab_size) # 등장 빈도수가 높은 상위 5개의 단어만 저장
print(vocab)
```

```python
# 높은 빈도 단어에 낮은 인덱스 부여
word_to_index = {}
i = 0
for (word, frequency) in vocab :
    i = i + 1
    word_to_index[word] = i
print(word_to_index)
```
