# 정규 표현식 이용(영어, 길이 1~2제거)

```python
import re
```

```python
text = "I was wondering if anyone out there could enlighten me on this car."
```

```python
# 길이가 1~2인 단어들을 정규 표현식을 이용하여 삭제
shortword = re.compile(r'\W*\b\w{1,2}\b')
print(shortword.sub('', text))
```

# 표제어 추출 도구

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
words = ['policy', 'doing', 'organization', 'have', 'going', 
        'love', 'lives', 'fly', 'dies', 'watched', 'has', 'starting']
```

```python
print('표제어 추출 전: ', words)
print('표제어 추출 후: ', [lemmatizer.lemmatize(word) for word in words])
```

```python
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize('dies', 'v'))
print(lemmatizer.lemmatize('watched', 'v'))
print(lemmatizer.lemmatize('has', 'v'))
```

# 어간 추출 알고리즘: 포터 알고리즘

```python
import nltk
nltk.download('punkt_tab')
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
```

```python
stemmer = PorterStemmer()
sentence = "This was not the map we found in Billy Bones's chest, but an accurate copy, complete in all things--names and heights and soundings--with the single exception of the red crosses and the written notes."
```

```python
tokenized_sentence = word_tokenize(sentence)
```

```python
print('어간 추출 전 :', tokenized_sentence)
print('어간 추출 후 :',[stemmer.stem(word) for word in
tokenized_sentence])
```

### 포터 알고리즘 vs 랭커스터 스태머(Lancaster Stemmer)

```python
from nltk.stem import PorterStemmer
from nltk.stem import LancasterStemmer
```

```python
porter_stemmer = PorterStemmer()
lancaster_stemmer = LancasterStemmer()
```

```python
words = ['policy', 'doing', 'organization', 'have', 'going',
'love', 'lives', 'fly', 'dies', 'watched', 'has',
'starting']
```

```python
print('어간 추출 전: ', words)
print('포터 스테머의 어간 추출 후: ',[porter_stemmer.stem(w)
for w in words])
print('랭커스터 스테머의 어간 추출 후: ',[lancaster_stemmer.stem(w) for w in words])
```

 단어 목록: policy, doing, organization, have, going, love, lives, fly, dies, watched, has, starting <br>
• Porter → ['polici','do','organ','have','go','love','live','fli','die','watch','ha','start'] <br>
• Lancaster → ['policy','doing','org','hav','going','lov','liv','fly','die','watch','has','start']

• 어간 추출: 빠르고 단순, 형태 왜곡·사전 비존재 형태 가능 <br>
• 표제어 추출: 품사/문맥 고려, 사전에 있는 형태로 반환 (정확도↑)
