```python
import nltk
nltk.download("punkt_tab")
```

```python
# 영어 코퍼스 토큰화 도구
from nltk.tokenize import word_tokenize
from nltk.tokenize import WordPunctTokenizer
import tensorflow as tf
text_to_word_sequence = tf.keras.preprocessing.text.text_to_word_sequence
```

단어 토큰화1 **관찰 포인트** <br>
• Don't → Do + n’t <br>
• Jone's → Jone + 's

```python
text = "Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."
```

```python
print('단어 토큰화1 :', word_tokenize(text))
```

단어 토큰화2 **관찰 포인트** <br>
• 구두점을 별도 토큰으로 분리 <br>
• Don't → Don + ' + t <br>
• Jone's → Jone + ' + s


```python
text = "Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."
```

```python
print('단어 토큰화2 :',WordPunctTokenizer().tokenize(text))
```

단어 토큰화3 **관찰 포인트** <br>
• 모든 알파벳 소문자화, 구두점 제거 <br>
• 아포스트로피 보존: "don't", "jone's"

```python
text = "Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."
```

```python
print('단어 토큰화3 :',text_to_word_sequence(text))
```

**표준 토큰화** 예: Penn Treebank <br>
• 규칙 1. 하이픈으로 구성된 단어는 하나로 유지 <br>
• 규칙 2. 아포스트로피 접어(doesn't 등)는 분리


• home-based → 한 토큰(규칙 1) <br>
• doesn't → does + n't로 분리(규칙 2)

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
```

```python
text = "Starting a home-based restaurant may be an ideal. it doesn't have a food chain or restaurant of their own."
```

```python
print('트리뱅크 워드토크나이저 :',tokenizer.tokenize(text))
```
